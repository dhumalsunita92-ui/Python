```python
import seaborn as sns # using for visualization barplot
import matplotlib.pyplot as plt  # using for customization(titile,labels)

plt.figure(figsize=(5,3))
sns.countplot(x="Risk_Category", data=a)
plt.title("Distribution of Risk Categories", color="red", fontweight="bold", loc="center")
plt.show()

sns.countplot(y="Job_Title", data=a)
plt.title("AI Effects on Job", color="red", fontweight="bold")   # Title
plt.xlabel("Count of Jobs", color="blue", fontsize=12)           # X-axis label
plt.ylabel("Job Titles", color="blue", fontsize=12)              # Y-axis label
plt.show()
```

```python
# Skills 
skills = [col for col in a.columns if "Skill" in col]
# Risk Category 
skill_matrix = a.groupby("Job_Title")[skills].mean()
# Heatmap
plt.figure(figsize=(12,6))
sns.heatmap(skill_matrix, annot=True, cmap="YlGnBu")
plt.title("Skill Importance by Risk Category", color="purple", fontweight="bold")
plt.show()

```
